### 📊 Understanding the Transaction Data Columns
Each row in the transaction dataset represents one ERC-20 token transfer event. The columns have the following meanings:
* `block_number` and `transaction_index`      
  Used to order transactions chronologically on the blockchain.
* `from_address`        
  The address that sends the token.
* `to_address`          
  The address that receives the token.
* `time_stamp`         
  The time of the transaction, recorded as a Unix timestamp (seconds since January 1, 1970, UTC).
* `contract_address`      
  Identifies which stablecoin or cryptocurrency is being transferred.
* `value`      
  The size of the transfer, representing the amount of capital moved.

### 🔎 Working with Time Information
#### The `time_stamp` Column
The `time_stamp` variable is stored as a Unix timestamp, which is not easily interpretable in its raw numeric form.     
To make the time information more readable (e.g. calendar dates and times), you may convert it into a standard datetime format using the code provided below.      
This step is optional but strongly recommended for visualization and analysis.

```python
"""
QUICK START: Drop-in replacement for your original visualization code

This integrates with your existing DatasetManager and Visualiser classes.
Just replace your original code block with this enhanced version.
"""

from src.DatasetManager import DatasetManager
from src.Visualiser import Visualiser
import pandas as pd
import numpy as np

# ============================================================================
# LOAD YOUR DATA (adjust to your actual data source)
# ============================================================================

# Option 1: If using DatasetManager
# dm = DatasetManager()
# token_df = dm.load_token_data('terra_ust')

# Option 2: Direct CSV load
# token_df = pd.read_csv('path/to/token_transfers.csv')

# For now, assuming you have token_df loaded with columns:
# - time_stamp (Unix timestamp)
# - value (transfer amount)
# - from_address, to_address


# ============================================================================
# DATA PREPARATION (ENHANCED)
# ============================================================================

print("Preparing data...")

# 1. Convert timestamp to datetime
token_df['Date'] = pd.to_datetime(token_df['time_stamp'], unit='s')

# 2. Sort by date (crucial!)
token_df = token_df.sort_values('Date').reset_index(drop=True)

# 3. Set as index for rolling operations
token_df_indexed = token_df.set_index('Date').sort_index()


# ============================================================================
# CALCULATE ENHANCED METRICS
# ============================================================================

print("Calculating crisis indicators...")

# Core Volume Metrics
token_df_indexed["Volume_24h"] = token_df_indexed["value"].rolling(
    window='24h', min_periods=1
).sum()

token_df_indexed["TxCount_24h"] = token_df_indexed["value"].rolling(
    window='24h', min_periods=1
).count()

# Whale Activity Indicators
token_df_indexed["AvgTxSize_24h"] = token_df_indexed["value"].rolling(
    window='24h', min_periods=1
).mean()

# Identify large transactions (>95th percentile)
large_tx_threshold = token_df_indexed["value"].quantile(0.95)
token_df_indexed["is_large"] = (token_df_indexed["value"] >= large_tx_threshold).astype(int)
token_df_indexed["LargeTxCount_24h"] = token_df_indexed["is_large"].rolling(
    window='24h', min_periods=1
).sum()

# Market Stress Indicators
token_df_indexed["TxVolatility_24h"] = token_df_indexed["value"].rolling(
    window='24h', min_periods=1
).std()

# Change rates (to detect acceleration)
token_df_indexed["Volume_Change_6h"] = token_df_indexed["Volume_24h"].pct_change(periods=36) * 100

# Reset index for Visualiser
token_df = token_df_indexed.reset_index()

print(f"✓ Data prepared: {len(token_df):,} transactions")
print(f"✓ Date range: {token_df['Date'].min()} to {token_df['Date'].max()}")


# ============================================================================
# VISUALISATION SETUP
# ============================================================================

# Initialize Visualiser
vis = Visualiser(token_df)

# Focus on crisis period + baseline
# Terra-Luna crisis: May 7-13, 2022
# Show from mid-April to capture baseline behavior
vis.focus("2022-04-15", "2022-05-31")

print("✓ Visualization focus set")


# ============================================================================
# DEFINE VISUALIZATION GRAPHS
# ============================================================================

graphs = [
    # Graph 1: Crisis Onset - When did confidence break?
    {
        "Title": "Crisis Onset: Volume and Transaction Surge",
        "Subtitle": "Sharp spikes indicate when mass redemptions began",
        "Traces": [
            {
                "Name": "24h Total Volume ($M)",
                "x": "Date",
                "y": lambda df: df["Volume_24h"] / 1e6,  # Convert to millions
                "Color": "crimson",
                "Type": "line"
            },
            {
                "Name": "24h Transaction Count",
                "x": "Date",
                "y": "TxCount_24h",
                "Color": "steelblue",
                "Type": "line",
                "yaxis": "y2"  # Secondary axis if Visualiser supports
            }
        ]
    },
    
    # Graph 2: Panic Propagation - Who moved first?
    {
        "Title": "Panic Propagation: Whale Activity",
        "Subtitle": "Large transactions spike before broad panic",
        "Traces": [
            {
                "Name": "Average Transaction Size ($)",
                "x": "Date",
                "y": "AvgTxSize_24h",
                "Color": "darkorange",
                "Type": "line"
            },
            {
                "Name": "Large Transactions (>95th %ile)",
                "x": "Date",
                "y": "LargeTxCount_24h",
                "Color": "darkred",
                "Type": "line",
                "yaxis": "y2"
            }
        ]
    },
    
    # Graph 3: Market Stress - How chaotic did it get?
    {
        "Title": "Market Stress: Transaction Size Volatility",
        "Subtitle": "Volatility spikes reveal market chaos",
        "Traces": [
            {
                "Name": "24h Transaction Volatility ($)",
                "x": "Date",
                "y": "TxVolatility_24h",
                "Color": "purple",
                "Type": "line"
            }
        ]
    },
    
    # Graph 4: Acceleration - How fast did it escalate?
    {
        "Title": "Crisis Acceleration: Volume Change Rate",
        "Subtitle": "Rate of change shows panic acceleration",
        "Traces": [
            {
                "Name": "6h Volume Change (%)",
                "x": "Date",
                "y": "Volume_Change_6h",
                "Color": "teal",
                "Type": "line"
            }
        ]
    }
]


# ============================================================================
# CREATE VISUALIZATIONS
# ============================================================================

print("\nGenerating visualizations...")

vis.plot_indicators(
    title="Terra-Luna Crisis: Confidence Breakdown and Panic Propagation",
    graphs=graphs
)

print("✓ Visualizations complete!")


# ============================================================================
# SUMMARY STATISTICS FOR YOUR REPORT
# ============================================================================

print("\n" + "="*70)
print("KEY FINDINGS FOR YOUR SLIDE DECK")
print("="*70)

# Find crisis onset
volume_baseline = token_df[token_df['Date'] < '2022-05-01']['Volume_24h'].mean()
volume_peak = token_df['Volume_24h'].max()
peak_date = token_df.loc[token_df['Volume_24h'].idxmax(), 'Date']

print(f"\n📊 CRISIS MAGNITUDE:")
print(f"   • Baseline daily volume: ${volume_baseline:,.0f}")
print(f"   • Peak daily volume: ${volume_peak:,.0f}")
print(f"   • Increase: {(volume_peak/volume_baseline - 1)*100:.1f}%")
print(f"   • Peak occurred: {peak_date}")

# Transaction patterns
txcount_baseline = token_df[token_df['Date'] < '2022-05-01']['TxCount_24h'].mean()
txcount_peak = token_df['TxCount_24h'].max()

print(f"\n🔄 TRANSACTION ACTIVITY:")
print(f"   • Baseline tx count: {txcount_baseline:.0f}/day")
print(f"   • Peak tx count: {txcount_peak:.0f}/day")
print(f"   • Increase: {(txcount_peak/txcount_baseline - 1)*100:.1f}%")

# Whale activity
large_tx_baseline = token_df[token_df['Date'] < '2022-05-01']['LargeTxCount_24h'].mean()
large_tx_peak = token_df['LargeTxCount_24h'].max()

print(f"\n🐋 WHALE EXODUS:")
print(f"   • Baseline large txs: {large_tx_baseline:.1f}/day")
print(f"   • Peak large txs: {large_tx_peak:.0f}/day")
print(f"   • Increase: {(large_tx_peak/large_tx_baseline - 1)*100:.1f}%")

# Market stress
volatility_baseline = token_df[token_df['Date'] < '2022-05-01']['TxVolatility_24h'].mean()
volatility_peak = token_df['TxVolatility_24h'].max()

print(f"\n⚡ MARKET STRESS:")
print(f"   • Baseline volatility: ${volatility_baseline:,.0f}")
print(f"   • Peak volatility: ${volatility_peak:,.0f}")
print(f"   • Increase: {(volatility_peak/volatility_baseline - 1)*100:.1f}%")

# First warning signals
crisis_start = token_df[token_df['Volume_24h'] > volume_baseline * 2]['Date'].min()
if pd.notna(crisis_start):
    print(f"\n🚨 EARLY WARNING:")
    print(f"   • First 2x volume spike: {crisis_start}")
    print(f"   • Days before peak: {(peak_date - crisis_start).days}")

print("\n" + "="*70)
print("\n💡 INTERPRETATION FOR SECTION A, Q1:")
print("\n   Confidence broke on [DATE] when volume increased [X]% and")
print("   transaction count surged [Y]%. Large players exited [Z] days")
print("   before peak panic, indicating informed actors detected instability")
print("   first. Market stress (volatility) increased [W]x, showing the")
print("   self-reinforcing nature of the run.")
print("\n" + "="*70)


# ============================================================================
# EXPORT DATA FOR FURTHER ANALYSIS
# ============================================================================

# Export hourly aggregates for cleaner slide charts
hourly_agg = token_df.set_index('Date').resample('1H').agg({
    'value': ['sum', 'mean', 'count', 'std']
}).reset_index()

hourly_agg.columns = ['Hour', 'Total_Volume', 'Avg_Tx', 'Tx_Count', 'Tx_Std']
hourly_agg.to_csv('/home/claude/terra_hourly_metrics.csv', index=False)

print("\n📁 Exported: terra_hourly_metrics.csv")
print("   Use this for cleaner visualizations in your slides")

print("\n✅ Analysis complete! Review the visualizations and summary statistics above.")
```
